```python
Code cell <undefined>
# %% [code]
import ultralytics
ultralytics.checks()

import torch
print(torch.backends.mps.is_available())

import os
os.getcwd()

Code cell <undefined>
# %% [code]
from ultralytics import YOLO
import time

start_time = time.time()

nepochs = 2
# Load a segmentation model
model = YOLO('yolov8n-seg.pt')

# Windows
data_yaml_path = os.path.join(os.getcwd(), 'data.yaml')

# Train using specific device
model.train(
    data=data_yaml_path,
    epochs=nepochs,
    imgsz=1024,
    device='mps'  # Use 'mps' for Apple GPU or 'cpu' for CPU or 'cuda' for NVIDIA GPU
)

results = model.val()  # Evaluate on the validation dataset
print(results)

model.save(f'Weed_classes_{nepochs}epochs.pt') # Save model to your computer

print("**** MODEL SAVED ****")

end_time = time.time()
print(f"Total time taken: {end_time - start_time} seconds")
```
